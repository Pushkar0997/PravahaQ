## 🟢 Layer 1: Plain-English Intuition

During flood or earthquake relief operations, a swarm of autonomous heavy-lift medical drones must deliver blood plasma and emergency comms payloads across isolated relief zones while sharing mobile **solar/battery-swap pads**.
- **Why Standard Quantum Heuristics Are Dangerous Here:** In life-critical emergency response, an algorithm that returns an invalid bitstring (*"Leave Relief Zone 2 with zero medical drops"* or *"Send 3 drones to a 1-pad charger simultaneously"*) is completely unacceptable.
- **How PravahaQ Guarantees Safety:** We combine **Backtracking Iterative Quantum Branch-and-Bound (`IQO-BnB`)** (which uses two-point spin correlations $\langle \hat{Z}_i \hat{Z}_j \rangle$ to avoid greedy traps) with our **RFC-0003 Deterministic Post-Quantum Feasibility Repair Engine**, guaranteeing **100.0% valid mission assignments**.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))
sys.path.insert(0, str(Path.cwd() / "src"))

from pravaha_q.domain import generate_coupled_problem
from pravaha_q.math_engine import VaisheshikaHUBOCompiler, hubo_to_ising, FeasibilityRepairEngine
from pravaha_q.algorithms import IQOBranchAndBoundSolver

drone_problem = generate_coupled_problem(
    num_nodes=3,
    num_tasks=4,
    num_power_states=1,
    seed=2026,
    name="Disaster-Drone-Swarm-15Q",
)
drone_hubo = VaisheshikaHUBOCompiler(enable_cubic_terms=True).compile(drone_problem)
drone_ising = hubo_to_ising(drone_hubo)

iqo = IQOBranchAndBoundSolver(classical_cutoff_vars=4, max_backtrack_depth=2, seed=2026)
iqo_res = iqo.solve(drone_ising)

mission = FeasibilityRepairEngine(drone_problem, drone_hubo).repair_and_decode(iqo_res.best_bitstring)
bases = ["Base-0 (Command-Mobile-Pad)", "Base-1 (Forward-Solar-Pad)", "Base-2 (Highland-Relay-Pad)"]

print("=== Mini-Project 12: Disaster-Relief Drone Swarm Mission Plan (IQO-BnB) ===")
print(f"Mission Scale              : {drone_problem.num_nodes} Mobile Charging Bases | {drone_problem.num_tasks} Medical Sorties | {drone_ising.num_qubits} Spin Variables")
print(f"IQO-BnB Branches Explored  : {iqo_res.branches_explored} Backtracking Branches")
print(f"IQO-BnB Optimal Energy     : {mission.repaired_energy:.4f} | Mission Bitstring: |{mission.repaired_bitstring}>")
print(f"100% Mission SLA Valid?    : {mission.is_feasible}\n")

print("Emergency Medical Sortie | Assigned Drone Base & Pad     | Payload Draw | Pad Status")
print("-" * 87)
for m, task in enumerate(drone_problem.workload.tasks):
    b_idx = mission.task_to_node[m]
    print(f"Medical-Sortie-{m}         | {bases[b_idx]:29s} |   {task.thermal_load_watts:6.2f} W   | CLEARED FOR LAUNCH")
```
